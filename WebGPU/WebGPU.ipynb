# WebGPU

WebGPU is a new web standard for accessing the GPU from web applications. It enables to use the power of the GPU (from client side) for high-performance computations. Usually, it is used for graphics rendering, but it can also be used for other types of computations, such as machine learning, physics simulations, and more.\
It reduces the workload of JavaScript and benifits ML inference on the web.

For example, TensorFlow.js can use WebGPU to accelerate machine learning inference in the browser. \
Another way to use WebGPU in the browser for machine is to use the Open Neural Network Exchange (ONNX) format, specifically with the ONNX Runtime Web library.
