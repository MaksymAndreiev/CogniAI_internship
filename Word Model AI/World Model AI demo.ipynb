# World Model AI
## What is World Model?
World Models are a neural networks, or system of them, that understands and simulates the dynamics of the world. It learns to predict future states of the environment based on past observations and actions. his predictions can be used by an agent to make desicions and plan actions to achieve goals.

Architecture:
- Encoder: processes multimodal inputs into a latent state.
- Predictor: learns a relationship between the states on the given action.
- Decoder: generates predictions of future states based on the current state and action.

## How it improves over transformer / current architecture?

Comparing with transformer-based models, World Models have several advantages. For example, they are more consistent, as they understand the underlying dynamics of the world, rather than just learning patterns in pixels (for example). \
Comparing with reinforcement learning without the world model, rl with world model can learn from experience without needing to interact with the environment, which can be more efficient and safer in some cases.
